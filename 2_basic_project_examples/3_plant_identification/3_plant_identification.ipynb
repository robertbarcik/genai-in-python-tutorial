# Plant ID

Point the app at a photo; it **looks** at the pixels and **names** the plant in one line. Same basic API call you already know, aimed at an image instead of text.

## Setup

Install the library and connect the client to your API key.

```python
# Colab installs what this notebook needs here. Running locally? `pip install -r requirements.txt` in the course folder already covers it.
%pip install -q openai==2.28.0

import os
from openai import OpenAI

# Key from Colab secrets, an environment variable, or a prompt, in that order.
try:
    from google.colab import userdata
    api_key = userdata.get("OPENAI_API_KEY")
except Exception:
    api_key = os.environ.get("OPENAI_API_KEY")
if not api_key:
    from getpass import getpass
    api_key = getpass("OpenAI API key: ")

client = OpenAI(api_key=api_key)
VISION_MODEL = "gpt-4o-mini"
```

## The photo you identify

Any public image URL works. Change it, re-run, get a new answer.

```python
from IPython.display import Image, display

photo_url = "https://commons.wikimedia.org/wiki/Special:FilePath/Sunflower_from_Silesia2.jpg?width=640"
display(Image(url=photo_url))
```

## The one call

This is the whole engine: one multimodal call that reads the photo and answers in plain text.

```python
identification = client.responses.create(
    model=VISION_MODEL,
    input=[
        {
            "role": "user",
            "content": [
                {"type": "input_text", "text": "Identify the plant in this photo. Reply with just the common name."},
                {"type": "input_image", "image_url": photo_url},
            ],
        }
    ],
).output_text

print(identification)
```
